# **Business Case Scenario**

You are working as a Junior Data Analyst at RetailKart, an online retail company that sells consumer products across India.

The marketing team spends money on:

1. TV Advertising
2. Social Media
3. Advertising
4. Influencer Marketing
5. Discounts

The management wants to understand:

**“Which factors impact product sales the most?”**

Your task is to build a Multiple Linear Regression Model to (a) predict Sales Revenue based on different marketing and business variables and (b) find out which factor has most impact on sales.

## **Objective**

Build a Multiple Regression model to:

1. Analyze relationships between variables
2. Predict Sales Revenue
3. Identify important business factors affecting sales
4. Evaluate model performance

The dataset contains 500 rows and the following columns:

| Column Name | Description |

| TV_Ads | Money spent on TV ads (₹) |
| Social_Media_Ads | Money spent on social media ads (₹) |
| Influencer_Marketing | Money spent on influencer marketing (₹) |
| Discount_Percentage | Discount offered (%) |

| Sales_Revenue | Total sales revenue (Target Variable) |

```python
# Import libraray
import pandas as pd
```

```python
# Read file
df = pd.read_csv('https://github.com/YBIFoundation/Assignmnet/raw/main/SalesPrediction.csv')
```

```python
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LinearRegression

X = df[['TV_Ads',
        'Social_Media_Ads',
        'Influencer_Marketing',
        'Discount_Percentage',
        'Website_Visits',
        'Customer_Rating']]

y = df['Sales_Revenue']

X_train, X_test, y_train, y_test = train_test_split(
    X, y,
    test_size=0.2,
    random_state=2529
)

model = LinearRegression()
model.fit(X_train, y_train)

print("Training complete!")
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
import numpy as np
import pandas as pd

y_pred = model.predict(X_test)

print("R2 Score:", r2_score(y_test, y_pred))
print("MAE:", mean_absolute_error(y_test, y_pred))
print("MSE:", mean_squared_error(y_test, y_pred))
print("RMSE:", np.sqrt(mean_squared_error(y_test, y_pred)))

importance = pd.DataFrame({
    'Feature': X.columns,
    'Coefficient': model.coef_
})

print("\nFeature Importance:")
print(importance.sort_values(by='Coefficient', ascending=False))
```
